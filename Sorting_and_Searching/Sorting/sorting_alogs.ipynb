# Bubble Sort

```python
import matplotlib.pyplot as plt 
import numpy as np

```

```python
arr = np.random.randint(0, 999, size=900)

n = len(arr)
operations = 0

for i in range(n-1): 
    swap = False
    for j in range(n-i-1):
        if arr[j] > arr[j+1]:
            arr[j], arr[j+1] = arr[j+1], arr[j]
            operations += 1
            swap=True
    if not swap:
        break        

print(f'Operation done to sort array: {operations}')            
```
