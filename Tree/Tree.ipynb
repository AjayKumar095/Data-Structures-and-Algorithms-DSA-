# Tree Data Structure

## Introduction

A Tree is a non-linear hierarchical data structure consisting of nodes connected by edges.

Unlike arrays and linked lists, data in a tree is organized in parent-child relationships.

Trees are widely used in:

- File systems
- Databases
- Search engines
- Compilers
- Artificial Intelligence
- Network routing

---

## Example Tree

```text
        A
      /   \
     B     C
    / \   / \
   D   E F   G
```

---

## Components of a Tree

### 1. Node

A node is the basic element of a tree.

Example:

```text
A, B, C, D, E, F, G
```

Each circle or value in the tree is a node.

---

### 2. Root Node

The topmost node of a tree is called the Root.

```text
        A
```

Here:

```text
Root = A
```

Every tree contains exactly one root node.

---

### 3. Parent Node

A node that has one or more children is called a parent node.

```text
A -> B, C
```

Here:

```text
A is the parent of B and C
```

---

### 4. Child Node

A node directly connected below another node is called a child.

```text
B and C are children of A
```

---

### 5. Sibling Nodes

Nodes having the same parent are called siblings.

```text
B and C
```

are siblings because both have parent A.

Similarly:

```text
D and E
F and G
```

are siblings.

---

### 6. Leaf Node

A node with no children is called a leaf node.

```text
D, E, F, G
```

are leaf nodes.

```text
Leaf Node:
left = None
right = None
```

---

### 7. Internal Node

Any node that has at least one child.

```text
A, B, C
```

are internal nodes.

---

### 8. Edge

A connection between two nodes.

```text
A ----- B
```

is one edge.

---

### 9. Subtree

A smaller tree inside a larger tree.

Example:

```text
      B
     / \
    D   E
```

This entire section is a subtree of A.

---

### 10. Depth

Depth is the number of edges from the root to a node.

Example:

```text
A -> B -> D
```

Depth of D = 2

---

### 11. Height

Height is the number of edges in the longest path from a node to a leaf.

Example:

```text
A
|
B
|
D
```

Height of A = 2

---

## Binary Tree

A Binary Tree is a tree in which every node has at most two children.

```python
class TreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
```

---

## Common Operations

- Insertion
- Deletion
- Searching
- Traversal

Traversal methods:

1. Preorder
2. Inorder
3. Postorder

These traversals use recursion and the call stack extensively.

---

## Summary

| Term | Meaning |
|--------|---------|
| Root | Top node |
| Parent | Node having children |
| Child | Node connected below a parent |
| Sibling | Nodes with same parent |
| Leaf | Node with no children |
| Edge | Connection between nodes |
| Depth | Distance from root |
| Height | Longest path to leaf |
| Subtree | Smaller tree inside a tree |

# Preorder Traversal

## Definition

Preorder Traversal visits nodes in the following order:

```text
Root → Left → Right
```

---

## Example Tree

```text
        R
      /   \
     A     B
    / \   / \
   C   D E   F
              \
               G
```

---

## Traversal Order

```text
R → A → C → D → B → E → F → G
```

Output:

```text
R,A,C,D,B,E,F,G
```

---

## Recursive Algorithm

```python
def preOrder(node):
    if not node:
        return

    print(node.data, end=",")

    preOrder(node.left)
    preOrder(node.right)
```

---

## How It Works

For node A:

```python
print(A)
preOrder(C)
preOrder(D)
```

The node is processed before its children.

---

## Call Stack Example

Start:

```python
preOrder(R)
```

Stack:

```text
R
```

Visit A:

```text
R
A
```

Visit C:

```text
R
A
C
```

Print sequence:

```text
R
A
C
```

When C finishes:

```text
R
A
```

Python returns to A and continues with D.

---

## Visualization

```text
Visit R
├── Visit A
│   ├── Visit C
│   └── Visit D
└── Visit B
    ├── Visit E
    └── Visit F
         └── Visit G
```

---

## Time Complexity

```text
O(n)
```

Every node is visited exactly once.

---

## Space Complexity

```text
O(h)
```

where h is the height of the tree.

Worst case:

```text
O(n)
```

---

## Use Cases

- Copying a tree
- Prefix expression evaluation
- Serializing trees
- Tree cloning

---

## Summary

```text
Preorder = Root → Left → Right
```

Output:

```text
R,A,C,D,B,E,F,G
```

````python
## pre-order tree traversal
node = root
def preOrder(node):
    if not node:
        return -1
    
    print(node.data, end=',')
    preOrder(node.left)
    preOrder(node.right)
    
preOrder(node)    
````

# Inorder Traversal

## Definition

Inorder Traversal visits nodes in the following order:

```text
Left → Root → Right
```

---

## Example Tree

```text
        R
      /   \
     A     B
    / \   / \
   C   D E   F
              \
               G
```

---

## Traversal Order

```text
C → A → D → R → E → B → F → G
```

Output:

```text
C,A,D,R,E,B,F,G
```

---

## Recursive Algorithm

```python
def inOrder(node):
    if not node:
        return

    inOrder(node.left)

    print(node.data, end=",")

    inOrder(node.right)
```

---

## How It Works

For node A:

```python
inOrder(C)

print(A)

inOrder(D)
```

The node is processed after the left subtree.

---

## Call Stack Example

Start:

```python
inOrder(R)
```

Move left until C:

```text
R
A
C
```

Print:

```text
C
```

Return to A.

Print:

```text
A
```

Visit D.

Print:

```text
D
```

Return to R.

Print:

```text
R
```

Continue with right subtree.

---

## Visualization

```text
C
↑
A
↓
D

↑
R

E
↑
B
↓
F
↓
G
```

---

## Time Complexity

```text
O(n)
```

---

## Space Complexity

```text
O(h)
```

---

## Important Property

For a Binary Search Tree (BST):

```text
Inorder Traversal produces sorted output.
```

Example:

```text
      5
     / \
    3   8
```

Output:

```text
3,5,8
```

---

## Summary

```text
Inorder = Left → Root → Right
```

Output:

```text
C,A,D,R,E,B,F,G
```

````python
## in-order tree travesal
def inOrder(node):
    if not node:
        return -1
    
    inOrder(node.left)
    print(node.data, end=',')
    inOrder(node.right)
    
inOrder(node)        
````

# Postorder Traversal

## Definition

Postorder Traversal visits nodes in the following order:

```text
Left → Right → Root
```

---

## Example Tree

```text
        R
      /   \
     A     B
    / \   / \
   C   D E   F
              \
               G
```

---

## Traversal Order

```text
C → D → A → E → G → F → B → R
```

Output:

```text
C,D,A,E,G,F,B,R
```

---

## Recursive Algorithm

```python
def postOrder(node):
    if not node:
        return

    postOrder(node.left)
    postOrder(node.right)

    print(node.data, end=",")
```

---

## How It Works

For node A:

```python
postOrder(C)

postOrder(D)

print(A)
```

The node is processed only after both children are completed.

---

## Call Stack Example

Start:

```python
postOrder(R)
```

Go left:

```text
R
A
C
```

C has no children.

Print:

```text
C
```

Return to A.

Visit D.

Print:

```text
D
```

Now both children of A are done.

Print:

```text
A
```

---

## Visualization

```text
C
D
A

E

G
F
B

R
```

Root is printed last.

---

## Time Complexity

```text
O(n)
```

---

## Space Complexity

```text
O(h)
```

---

## Use Cases

- Delete a tree
- Free memory
- Evaluate expression trees
- Calculate subtree values

---

## Why Tree Deletion Uses Postorder

To safely delete:

```text
Delete children first
Delete parent last
```

Example:

```text
C
D
A
```

If A were deleted first, access to C and D would be lost.

---

## Summary

```text
Postorder = Left → Right → Root
```

Output:

```text
C,D,A,E,G,F,B,R
```

Root is always processed after both subtrees are completed.

````python
## post-order tree travesal
def psotOrder(node):
    if not node:
        return -1
    
    psotOrder(node.left)
    psotOrder(node.right)
    print(node.data, end=',')
    
psotOrder(node) 
````
